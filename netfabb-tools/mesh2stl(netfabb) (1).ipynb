```python
import trimesh
import numpy as np
import os
def get_filelist(path, include=None, exclude=None):
    Filelist = []
    for home, dirs, files in os.walk(path):
        for filename in files:
            include_criteria=True
            exclude_criteria=True
            
            if isinstance(include, str):
                if include not in filename:
                    include_criteria=False

            elif isinstance(include, list):
                for i in include:
                    if i not in filename:
                        include_criteria=False
                        continue

            if isinstance(exclude, str):
                if exclude in filename:
                    exclude_criteria=False

            elif isinstance(exclude, list):
                for e in exclude:
                    if e in filename:
                        exclude_criteria=False
                        continue

            if include_criteria==True and exclude_criteria==True:
                Filelist.append(home+'/'+filename)

    return Filelist
```

```python
path = r"C:\Users\XuanLiang\Documents\Netfabb_Fusion360_data\OBJ"
obj_files = get_filelist(path, include='obj')
```

```python
stl_dir = r"C:\Users\XuanLiang\Documents\Netfabb_Fusion360_data\STL"
for i, obj_file in enumerate(obj_files[:]):
    name = obj_file.split('/')[-1].split('.')[0]
    obj_start = trimesh.load_mesh(obj_file)
    obj_temp = obj_start.copy()
    obj_temp.apply_scale(60)
    
    v = np.array(obj_temp.vertices)
    # x = v[:,0]
    # y = v[:,1]
    z = v[:,2]
    obj_temp.apply_translation((0,0,-z.min()))
    # v = np.array(obj_temp.vertices)
    # z = v[:,2]
    # print(z.min())
    stl_path = stl_dir+'/'+name+'.stl'
    obj_temp.export(stl_path)
    print(f"Progress: {i/len(obj_files)*100.:.2f}%", end="\r")
```
